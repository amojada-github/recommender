```python
# Step 1: Specify the Order of Values for Each Column
# Step 2: Use pandas.map() to Map the Orders to Values

quality_dict = {
    'Ex': 4,
     'Gd': 3,
     'TA': 2,
     'Fa': 1,
     'Po': 0
}

cleaned_train_data['KitchenQual'] = cleaned_train_data['KitchenQual'].map(quality_dict)
cleaned_train_data['ExterQual'] = cleaned_train_data['ExterQual'].map(quality_dict)
cleaned_train_data['ExterCond'] = cleaned_train_data['ExterCond'].map(quality_dict)
cleaned_train_data['HeatingQC'] = cleaned_train_data['HeatingQC'].map(quality_dict)

street_dict = {
    'Grvl': 0,
    'Pave': 1
}

cleaned_train_data['Street'] = cleaned_train_data['Street'].map(street_dict)


utilities_dict = {
    'AllPub': 0,
    'NoSeWa': 1
}
cleaned_train_data['Utilities'] = cleaned_train_data['Utilities'].map(utilities_dict)

central_air_dict = {
    'N':0 ,
    'Y':1
    }

cleaned_train_data['CentralAir'] = cleaned_train_data['CentralAir'].map(central_air_dict)

y_n_p_dict = {
    'N':0 ,
    'P':1 ,
    'Y':2 ,
    }

cleaned_train_data['PavedDrive'] = cleaned_train_data['PavedDrive'].map(y_n_p_dict)

lotshape_dict = {
    'Reg': 3,
    'IR1': 2,
    'IR2': 1,
    'IR3': 0
    }

cleaned_train_data['LotShape'] = cleaned_train_data['LotShape'].map(lotshape_dict)

functional_dict = {
    'Sev': 0,
    'Maj2': 1,
    'Maj1': 2,
    'Mod': 3,
    'Min2': 4,
    'Min1': 5,
    'Typ': 6
 }

cleaned_train_data['Functional'] = cleaned_train_data['Functional'].map(functional_dict)
```

Nominal Categorical Features Encoding

```python
for cat_ft in nominal_cat_ft:
    print(cat_ft, ': ', cleaned_train_data[cat_ft].unique())
```

One-Hot (Nominal) Encoding

```python
for cat_ft in nominal_cat_ft + ['MSSubClass', 'MoSold']: # numerical_categoricals_to_encode
    dummies = pd.get_dummies(cleaned_train_data[cat_ft], prefix=cat_ft, dtype=int)
    cleaned_train_data = pd.concat([cleaned_train_data, dummies], axis=1)
    cleaned_train_data.drop(cat_ft, axis=1, inplace=True)
```

```python
features_to_scale = [
    'LotArea',
    'YearBuilt',
    'YearRemodAdd',
    'BsmtFinSF1',
    'BsmtFinSF2',
    'BsmtUnfSF',
    'TotalBsmtSF',
    '1stFlrSF',
    '2ndFlrSF',
    'LowQualFinSF',
    'GrLivArea',
    'BsmtFullBath',
    'BsmtHalfBath',
    'FullBath',
    'HalfBath',
    'BedroomAbvGr',
    'KitchenAbvGr',
    'TotRmsAbvGrd',
    'Fireplaces',
    'GarageCars',
    'GarageArea',
    'WoodDeckSF',
    'OpenPorchSF',
    'EnclosedPorch',
    '3SsnPorch',
    'ScreenPorch',
    'PoolArea',
    'MiscVal',
    'YrSold'
]

ordinal_features_to_scale = [
    'LotShape',
    'OverallQual',
    'OverallCond',
    'ExterQual',
    'ExterCond',
    'HeatingQC',
    'KitchenQual',
    'Functional',
    'PavedDrive'
]

features_to_scale = features_to_scale + ordinal_features_to_scale
```

```python
X_mean = X_train[features_to_scale].mean()
X_std = X_train[features_to_scale].std()

X_train_scaled = X_train.copy()
X_test_scaled = X_test.copy()

X_train_scaled[features_to_scale] = (X_train[features_to_scale] - X_mean) / X_std
X_test_scaled[features_to_scale] = (X_test[features_to_scale] - X_mean) / X_std
```

```python
# Gradient Descent

best_w = None
best_b = None
best_cost = None


# Leveraging Gradient Descent to find the best model parameters
m, n = X_train_scaled.shape
dj_dw, dj_db = np.zeros(n), 0.0
w, b = np.zeros(n), 0.0
learning_rate = 0.01
for iter in range(10):

    w = w - learning_rate * dj_dw
    b = b - learning_rate * dj_db

    error_sqrd = 0.
    dj_dw = np.zeros(n)
    dj_db = 0.
    for i in range(m):
        y_hat = np.dot(w, X_train_scaled.iloc[i].values) + b
        error_sqrd += (y_hat - y_train.iloc[i])**2

        for j in range(n):
            dj_dw[j] += (y_hat - y_train.iloc[i]) * X_train_scaled.iloc[i, j]

        dj_db += (y_hat - y_train.iloc[i])

    cost = error_sqrd / (2 * m)
    dj_dw = dj_dw / m
    dj_db = dj_db / m

    print(iter, cost) #, w, b) 

    if best_cost is None or cost < best_cost:
        best_cost = cost
        best_w = w
        best_b = b

predictions = []
for i in range(m):
    predictions.append((np.dot(best_w, X_train_scaled.iloc[i].values) + best_b ))

print(predictions)
```

```python
sse = 0
test_y_hat = []
for i in range(len(X_test_scaled)):
    y_hat = np.dot(best_w, X_test_scaled.iloc[i].values) + best_b 
    sse += (y_hat - y_test.iloc[i])**2
    test_y_hat.append(y_hat)

m = len(X_test_scaled)
cost = ( 1 / (2 * m)) * sse

print('cost: ', cost)
```
