#### Feature Engineering

```python
X_train
```

```python
X_train = np.c_[X_train, X_train[:,1]**2]
X_train
```

#### Feature Scaling

```python
mu = np.mean(X_train, axis=0)
sigma = np.std(X_train, axis=0)

X_train_norm = (X_train - mu) / sigma
np.ptp(X_train, axis=0), np.ptp(X_train_norm, axis=0)
```

#### Implementing Gradient Descent

```python
m, n = X_train_norm.shape

w = np.zeros(n)
b = 0.0

dj_dw = np.zeros(n)
dj_db = 0.0

learning_rate = 0.1
for iter in range(1000):

    w = w - learning_rate * dj_dw
    b = b - learning_rate * dj_db

    error_sqrd = 0
    dj_dw = np.zeros(n)
    dj_db = 0.0
    for i in range(m):
        y_hat = np.dot(w, X_train_norm[i]) + b

        error_sqrd += (y_hat - y_train[i])**2

        for j in range(n):
            dj_dw[j] += (y_hat - y_train[i]) * X_train_norm[i, j]
        dj_db += (y_hat - y_train[i])

    cost = error_sqrd / (2 * m)
    dj_dw = dj_dw / m
    dj_db = dj_db / m


    print(iter, w, b, cost)


predictions = []
for i in range(m):
    predictions.append(np.dot(w, X_train_norm[i]) + b)

print(predictions)
```
