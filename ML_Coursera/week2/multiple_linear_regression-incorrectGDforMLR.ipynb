```python
import numpy as np
```

```python
X_train_raw = np.array([[2104, 5, 1, 45], [1416, 3, 2, 40], [852, 2, 1, 35]])
y_train = np.array([460, 232, 178])

mean = X_train_raw.mean(axis=0)
std = X_train_raw.std(axis=0)

X_train = (X_train_raw - mean) / std
```

```python
# Leveraging Gradient Descent to find the best model parameters
m = len(X_train)
dj_dw_1, dj_dw_2, dj_dw_3, dj_dw_4, dj_db = 0, 0, 0, 0, 0
w_1, w_2, w_3, w_4, b = 0, 0, 0, 0, 0
learning_rate = 0.1
for iter in range(500):

    w_1 = w_1 - learning_rate * dj_dw_1
    w_2 = w_2 - learning_rate * dj_dw_2
    w_3 = w_3 - learning_rate * dj_dw_3
    w_4 = w_4 - learning_rate * dj_dw_4
    b = b - learning_rate * dj_db

    error_sqrd = 0
    dj_dw_val_1, dj_dw_val_2, dj_dw_val_3, dj_dw_val_4, dj_db_val = 0, 0, 0, 0, 0
    for i in range(m):
        y_hat = w_1 * X_train[i, 0] + w_2 * X_train[i, 1] + w_3 * X_train[i, 2] + w_4 * X_train[i, 3] + b

        error_sqrd += (y_hat - y_train[i])**2

        dj_dw_val_1 += (y_hat - y_train[i]) * X_train[i, 0]
        dj_dw_val_2 += (y_hat - y_train[i]) * X_train[i, 1]
        dj_dw_val_3 += (y_hat - y_train[i]) * X_train[i, 2]
        dj_dw_val_4 += (y_hat - y_train[i]) * X_train[i, 3]

        dj_db_val += (y_hat - y_train[i])


    cost = error_sqrd / (2 * m)
    dj_dw_1 = dj_dw_val_1 / m
    dj_dw_2 = dj_dw_val_2 / m
    dj_dw_3 = dj_dw_val_3 / m
    dj_dw_4 = dj_dw_val_4 / m
    dj_db = dj_db_val / m

    print(iter, w_1, w_2, w_3, w_4, b, cost) 

predictions = []
for i in range(m):
    predictions.append((w_1 * X_train[i, 0] + w_2 * X_train[i, 1] + w_3 * X_train[i, 2] + w_4 * X_train[i, 3] + b ))

print(predictions)
```

```python
# Leveraging Gradient Descent to find the best model parameters
m, n = X_train.shape
dj_dw, dj_db = np.zeros(n), 0
w, b = np.zeros(n), 0
learning_rate = 0.1
for iter in range(500):

    w = w - learning_rate * dj_dw
    b = b - learning_rate * dj_db

    error_sqrd = 0
    dj_dw = np.zeros(n)
    dj_db = 0.
    for i in range(m):
        y_hat = np.dot(w, X_train[i]) + b
        error_sqrd += (y_hat - y_train[i])**2

        for j in range(n):
            dj_dw[j] += (y_hat - y_train[i]) * X_train[i, j]

        dj_db += (y_hat - y_train[i])


    cost = error_sqrd / (2 * m)
    dj_dw = dj_dw / m
    dj_db = dj_db / m

    print(iter, w, b, cost) 

predictions = []
for i in range(m):
    predictions.append((np.dot(w, X_train[i]) + b ))

print(predictions)
```
