```python
split_at = int(0.8 * len(train_data))

train_rows = train_data.iloc[:split_at]
validation_rows = train_data.iloc[split_at:]


x_train = train_rows['GrLivArea'].to_numpy() / 1000
y_train = train_rows['SalePrice'].to_numpy() / 1000
```

```python
x_train = x_train[:3]
y_train = y_train[:3]
```

```python
# I want to leverage the gradient descent to find the best parameters and make the most accurate predictions

# Since this is a single variable linear regression model, I have two choices
# 1 Direct calculation of w, b
# 2 Leverage Gradient Descent to find the w, b


# 1 Direct Calculation
# step 1: calculate the partial derivative of J relative to w and b
# step 2: plug one partial derivative into the other and solve two equations to find both w, b
# step 3: make predictions
        # Predictions can be made in two ways
            # 1 Using a for loop
            # 2 Using vectorization
            

import numpy as np


x_mean = x_train.mean()
y_mean = y_train.mean()

best_w_dir = np.sum((x_train - x_mean) * (y_train - y_mean)) / np.sum((x_train - x_mean)**2)
best_b_dir = y_mean - best_w_dir * x_mean

# Making predictions using vectorization
predictions = best_w_dir * x_train + best_b_dir # ~ y_train

print('prediction - via vectorization', predictions)
```

```python

# Leveraging Gradient Descent to find the best model parameters

m = len(x_train)
dj_dw, dj_db = 0, 0
w, b = 0, 0
learning_rate = 0.01
for iter in range(2000):

    w = w - learning_rate * dj_dw
    b = b - learning_rate * dj_db

    error_sqrd = 0
    dj_dw_val, dj_db_val = 0, 0
    for i in range(m):
        y_hat = w * x_train[i] + b
        error_sqrd += (y_hat - y_train[i])**2

        dj_dw_val += (y_hat - y_train[i]) * x_train[i]
        dj_db_val += (y_hat - y_train[i])


    cost = error_sqrd / (2 * m)
    dj_dw = dj_dw_val / m
    dj_db = dj_db_val / m

    print(iter, w, b, cost) 

predictions = []

for i in range(m):
    predictions.append(w * x_train[i] + b)

print(predictions)
```

```python
def func(x, w, b):
    return w * x + b

def squared_error(y_hat, y): 
    return (y_hat - y)**2
```

```python
best_w = None
best_b = None
best_cost = None

learning_rate = 0.01
w, b = 0, 0
for step in range(0,100000):

    if step != 0:
        w = w - learning_rate * dj_dw
        b = b - learning_rate * dj_db

    y_hat = []
    for x in x_train:
        y_hat.append(func(x, w, b))


    sum_of_squared_errors = 0
    dj_dw = 0
    dj_db = 0
    for i in range(len(y_hat)):
        sum_of_squared_errors += squared_error(y_hat[i], y_train[i])
        dj_dw += (y_hat[i] - y_train[i]) * x_train[i]
        dj_db += (y_hat[i] - y_train[i])

    m = len(x_train)
    cost = ( 1/ (2 * m) ) * sum_of_squared_errors
    dj_dw = ( 1/ m ) * dj_dw
    dj_db = ( 1/ m ) * dj_db

    if best_cost is None or cost < best_cost:
        best_cost = cost
        best_w = w
        best_b = b

    if step % 1000 == 0:
        print('step: ', step, w, b, 'cost: ', cost, 'dj_dw', dj_dw, 'dj_db', dj_db)
    if best_cost is not None and (abs(dj_dw) < 1e-6 and abs(dj_db) < 1e-6):
        break

print('Best Parameters are:', step, w, b, cost)
```

```python
x_valid = validation_rows['GrLivArea'].to_numpy() / 1000
y_valid = validation_rows['SalePrice'].to_numpy() / 1000

sse = 0
valid_y_hat = []
for i in range(len(x_valid)):
    y_hat = func(x_valid[i], best_w, best_b)
    sse += (y_hat - y_valid[i])**2
    valid_y_hat.append(y_hat * 1000)

m = len(x_valid)
cost = ( 1 / (2 * m)) * sse

print('cost: ', cost)
```
